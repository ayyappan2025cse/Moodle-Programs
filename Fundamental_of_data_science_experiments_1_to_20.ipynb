```python

from cryptography.fernet import Fernet

# Generate a secret key
key = Fernet.generate_key()

# Create Fernet object
cipher = Fernet(key)

# Original sensitive data
data = "Name: Arun, Account Number: 123456"

print("Original Data:", data)

# Encrypt the data
encrypted_data = cipher.encrypt(data.encode())
print("Encrypted Data:", encrypted_data)

# Decrypt the data
decrypted_data = cipher.decrypt(encrypted_data).decode()
print("Decrypted Data:", decrypted_data)
```

```python

# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# Upload CSV file
uploaded = files.upload()

# Read the dataset
filename = next(iter(uploaded))
df = pd.read_csv(filename)

# Display the dataset
print("First 5 Rows:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nDataset Shape:")
print(df.shape)

# Data preprocessing
print("\nMissing Values:")
print(df.isnull().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Convert Date column into date format
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')

# Convert numeric columns
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')

# Fill missing numeric values with column averages
df['Sales'] = df['Sales'].fillna(df['Sales'].mean())
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].mean())

# Remove rows with missing important values
df = df.dropna(subset=['Date', 'Product', 'Region'])

print("\nCleaned Dataset:")
print(df.head())

print("\nDataset After Preprocessing:")
print(df.info())

# Data analysis
print("\nTotal Sales:", df['Sales'].sum())
print("Total Quantity Sold:", df['Quantity'].sum())

print("\nSales by Product:")
print(df.groupby('Product')['Sales'].sum())

print("\nSales by Region:")
print(df.groupby('Region')['Sales'].sum())

# Visualization 1: Sales by Product
df.groupby('Product')['Sales'].sum().plot(
    kind='bar', title='Total Sales by Product'
)
plt.xlabel('Product')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()

# Visualization 2: Sales by Region
df.groupby('Region')['Sales'].sum().plot(
    kind='pie', autopct='%1.1f%%',
    title='Sales Distribution by Region'
)
plt.ylabel('')
plt.show()

# Visualization 3: Sales over Time
df.groupby('Date')['Sales'].sum().sort_index().plot(
    kind='line', marker='o', title='Sales Over Time'
)
plt.xlabel('Date')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Visualization 4: Quantity Sold by Product
df.groupby('Product')['Quantity'].sum().plot(
    kind='bar', title='Quantity Sold by Product'
)
plt.xlabel('Product')
plt.ylabel('Quantity')
plt.tight_layout()
plt.show()

# Save cleaned dataset
df.to_csv('cleaned-sales-data.csv', index=False)

print("\nCleaned dataset saved successfully!")

# Download cleaned dataset
files.download('cleaned-sales-data.csv')

```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
population = np.random.normal(loc=170, scale=10, size=10000)

print(f"Population Mean: {np.mean(population):.2f}")
print(f"Population Std Dev: {np.std(population):.2f}")

sample_size = 30
sample_mean = np.mean(np.random.choice(population, size=sample_size))
print(f"Single Sample Mean (n={sample_size}): {sample_mean:.2f}")

num_samples = 1000
sample_means = [np.mean(np.random.choice(population, size=sample_size)) for _ in range(num_samples)]

print(f"Sampling Distribution Mean: {np.mean(sample_means):.2f}")
print(f"Sampling Distribution Std Error: {np.std(sample_means):.2f}")
print(f"Theoretical Standard Error: {np.std(population) / np.sqrt(sample_size):.2f}")

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(population, bins=30, color='skyblue', edgecolor='black')
plt.title("Population Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(sample_means, bins=30, color='salmon', edgecolor='black')
plt.title(f"Sampling Distribution of Means (n={sample_size})")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from statsmodels.stats.weightstats import ztest

data = np.array([148, 153, 146, 155, 152, 149, 151, 147, 154, 150,
                 152, 148, 153, 149, 151, 156, 145, 150, 152, 148])

hypothesized_mean = 150

z_stat, p_value = ztest(data, value=hypothesized_mean)

alpha = 0.05

print(f"Sample Mean: {np.mean(data):.2f}")
print(f"Z-statistic: {z_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < alpha:
    print("Reject Null Hypothesis: Average weight differs significantly from 150 grams.")
else:
    print("Fail to Reject Null Hypothesis: No significant difference from 150 grams.")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = np.array([12, 15, 14, 10, 13, 12, 14, 100, 11, 15, 13, 12, 11, 14, 105, 12, 13, 10])
df = pd.DataFrame(data, columns=['Values'])

q1 = df['Values'].quantile(0.25)
q3 = df['Values'].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

iqr_outliers = df[(df['Values'] < lower_bound) | (df['Values'] > upper_bound)]['Values'].tolist()

mean = df['Values'].mean()
std = df['Values'].std()

z_scores = (df['Values'] - mean) / std
z_outliers = df[abs(z_scores) > 3]['Values'].tolist()

print(f"IQR Lower Bound: {lower_bound:.2f}, Upper Bound: {upper_bound:.2f}")
print("Outliers detected using IQR method:", iqr_outliers)

print(f"\nMean: {mean:.2f}, Standard Deviation: {std:.2f}")
print("Outliers detected using Z-score method (|Z| > 3):", z_outliers)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.boxplot(y=df['Values'], color='skyblue')
plt.title("Boxplot for Outlier Detection (IQR Method)")

plt.subplot(1, 2, 2)
plt.scatter(df.index, df['Values'], color='blue', label='Data Points')
plt.scatter(df[abs(z_scores) > 3].index, z_outliers, color='red', label='Z-score Outliers')
plt.axhline(mean, color='green', linestyle='--', label='Mean')
plt.title("Scatter Plot for Outlier Detection")
plt.xlabel("Index")
plt.ylabel("Values")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
df=pd.read_csv('/content/gemini-code-1791908688839.txt')
X=df[['Age','Annual_Income']]
y=df['Purchased']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
knn_unscaled=KNeighborsClassifier(n_neighbors=3)
knn_unscaled.fit(X_train,y_train)
acc_unscaled=accuracy_score(y_test,knn_unscaled.predict(X_test))
minmax=MinMaxScaler()
X_train_minmax=minmax.fit_transform(X_train)
X_test_minmax=minmax.transform(X_test)
knn_minmax=KNeighborsClassifier(n_neighbors=3)
knn_minmax.fit(X_train_minmax,y_train)
acc_minmax=accuracy_score(y_test,knn_minmax.predict(X_test_minmax))
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
knn_standard=KNeighborsClassifier(n_neighbors=3)
knn_standard.fit(X_train_scaled,y_train)
acc_standard=accuracy_score(y_test,knn_standard.predict(X_test_scaled))
print(f"Accuracy without Scaling: {acc_unscaled*100:.2f}%")
print(f"Accuracy with MinMax Scaling: {acc_minmax*100:.2f}%")
print(f"Accuracy with Standardization: {acc_standard*100:.2f}%")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
df=pd.read_csv('/content/gemini-code-1791909468780.txt')
X=df[['Age','Estimated_Salary']]
y=df['Purchased']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
knn=KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train_scaled,y_train)
y_pred=knn.predict(X_test_scaled)
print(f"Accuracy: {accuracy_score(y_test,y_pred)*100:.2f}%")
print("Confusion Matrix:\n",confusion_matrix(y_test,y_pred))
```

```python
import io
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix

csv_data = """Tenure_Months,Monthly_Charges,Churn
1,70,1
12,20,0
3,85,1
60,100,0
2,65,1
24,45,0
48,110,0
5,90,1
36,55,0
8,75,1"""

df = pd.read_csv(io.StringIO(csv_data))

X = df[['Tenure_Months','Monthly_Charges']]
y = df['Churn']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model = LogisticRegression()
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)

print(f"Accuracy: {accuracy_score(y_test, y_pred)*100:.2f}%")
print(f"Precision: {precision_score(y_test, y_pred):.2f}")
print(f"Recall: {recall_score(y_test, y_pred):.2f}")
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
import io
import pandas as pd
from sklearn.cluster import KMeans

csv_data = """Annual_Income,Spending_Score
15,39
15,81
16,6
16,77
17,40
17,76
18,94
19,3
19,72
20,14"""

df = pd.read_csv(io.StringIO(csv_data))
X = df[['Annual_Income', 'Spending_Score']]

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X)

print("Cluster Centroids:\n", kmeans.cluster_centers_)
print("\nClustered Data:\n", df)
print(f"\nInertia (WCSS): {kmeans.inertia_:.2f}")
```
